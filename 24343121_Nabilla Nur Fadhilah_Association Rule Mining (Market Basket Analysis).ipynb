```python
import pandas as pd
import numpy as np
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

# Membaca kedua dataset
df = pd.read_csv('transactions_arm_praktikum.csv')
df_onehot = pd.read_csv('transactions_arm_praktikum_onehot.csv')

# Mengubah kolom Items menjadi list transaksi
transactions = df['Items'].apply(
    lambda x: [item.strip() for item in x.split(',')]
).tolist()

# Memastikan dataset One-Hot Encoding terbaca sebagai boolean
df_onehot = df_onehot.astype(bool)

# Menampilkan jumlah transaksi dan 5 transaksi pertama
print('Jumlah transaksi:', len(transactions))

for i, t in enumerate(transactions[:5], 1):
    print(f'T{i}: {t}')
```

Langkah awal adalah mengimpor pustaka yang dibutuhkan: pandas dan numpy untuk manipulasi data, serta TransactionEncoder, apriori, fpgrowth, dan association_rules dari mlxtend untuk penambangan aturan asosiasi. Data dibaca dari dua berkas CSV, yaitu transactions_arm_praktikum.csv (satu baris per transaksi, item dipisahkan koma) dan transactions_arm_praktikum_onehot.csv (versi one-hot). Kolom Items kemudian dipecah dengan split(',') dan dirapikan dengan strip() sehingga menjadi list of list, yaitu satu daftar item untuk setiap transaksi. Dataset one-hot diubah ke tipe boolean dengan astype(bool).


Penjelasan output. Output menunjukkan bahwa dataset berisi 60 transaksi. Lima transaksi pertama adalah T1: Gula, Kopi, Susu; T2: Gula, Kopi, Susu; T3: Mentega, Roti, Susu, Telur; T4: Mie Instan; dan T5: Keju, Roti, Susu, Telur. Jumlah item per transaksi berbeda-beda, sesuai karakteristik data transaksi, dan hasil ini sama dengan contoh pada modul.


```python
from collections import Counter
import matplotlib.pyplot as plt

# Menghitung frekuensi kemunculan setiap item
cnt = Counter()

for t in transactions:
    cnt.update(t)

# Mengatur urutan item jika frekuensinya sama
urutan_item = [
    'Susu', 'Mie Instan', 'Roti', 'Mentega',
    'Sabun', 'Telur', 'Gula', 'Shampo',
    'Kopi', 'Kecap', 'Keju', 'Teh'
]

# Mengurutkan berdasarkan frekuensi tertinggi
items_sorted = sorted(
    cnt.keys(),
    key=lambda item: (-cnt[item], urutan_item.index(item))
)

# Menampilkan frekuensi dan support
for item in items_sorted:
    freq = cnt[item]
    support = freq / len(transactions)
    print(f'{item:12s}: {freq:2d}  (support = {support:.3f})')

# Membuat grafik frekuensi item
frekuensi = [cnt[item] for item in items_sorted]

plt.figure(figsize=(10, 5))
bars = plt.bar(items_sorted, frekuensi, color='#202B68')

# Menampilkan angka di atas setiap batang
for bar in bars:
    tinggi = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        tinggi + 0.2,
        f'{int(tinggi)}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.title('Frekuensi Kemunculan Item (dari 60 Transaksi)')
plt.xlabel('Item')
plt.ylabel('Jumlah Transaksi')
plt.xticks(rotation=30, ha='right')
plt.ylim(0, 22)
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

Sebelum Apriori dijalankan, frekuensi kemunculan tiap item dihitung terlebih dahulu untuk mengetahui item mana yang populer dan mana yang jarang dibeli. Objek Counter dari modul collections dibuat, lalu setiap transaksi dimasukkan melalui cnt.update(t) sehingga Counter menghitung berapa kali tiap item muncul. Daftar urutan_item digunakan sebagai aturan pengurutan sekunder. Jika dua item memiliki frekuensi sama, misalnya Mie Instan dan Roti yang sama-sama 18, urutannya mengikuti daftar ini agar hasilnya konsisten dan sesuai dengan modul. Item diurutkan dari frekuensi tertinggi dengan sorted() dan key=lambda item: (-cnt[item], urutan_item.index(item)). Support tiap item dihitung dengan rumus freq / len(transactions), yaitu jumlah transaksi yang memuat item dibagi total 60 transaksi, lalu dicetak dengan tiga angka desimal.
Hasil hitungan ini kemudian divisualisasikan dengan plt.bar() pada ukuran gambar 10×5. Angka frekuensi ditampilkan di atas setiap batang memakai plt.text(). Label sumbu x diputar 30° (rotation=30) agar nama item tidak saling menimpa, dan plt.ylim(0, 22) memberi ruang di atas batang tertinggi agar angkanya tidak terpotong.


Penjelasan output : Susu menjadi item paling sering dibeli dengan 19 transaksi (support 0,317). Berikutnya Mie Instan dan Roti dengan 18 transaksi (0,300), lalu Mentega 17 (0,283), Sabun dan Telur 15 (0,250), Gula 14 (0,233), Shampo dan Kopi 13 (0,217), serta Kecap 12 (0,200). Dua item paling jarang adalah Keju dengan 6 transaksi (0,100) dan Teh dengan 4 transaksi (0,067).


Penjelasan grafik : Diagram batang memperlihatkan pola yang sama secara visual. Sepuluh item pertama frekuensinya relatif berdekatan (12 sampai 19), sedangkan Keju dan Teh turun tajam. Informasi ini penting untuk menentukan min_support. Nilai yang terlalu tinggi akan membuang hampir semua item, dan karena support tertinggi hanya 0,317, nilai di atas itu tidak menghasilkan itemset sama sekali. Karena itu dipilih min_support = 0,15, yang akan membuang Keju dan Teh tetapi mempertahankan item lainnya.


```python
# Mengubah transaksi menjadi One-Hot Encoding
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)

# Membuat DataFrame dari hasil encoding
df = pd.DataFrame(te_ary, columns=te.columns_)

# Menampilkan ukuran dan 5 baris pertama
print('Ukuran matriks one-hot:', df.shape)
print(df.iloc[:5, :6])
```

 Fungsi apriori() dan fpgrowth() pada mlxtend tidak dapat memproses data berbentuk list of list, melainkan membutuhkan matriks True/False. Karena itu TransactionEncoder digunakan untuk mengubah format data. Method fit(transactions) mempelajari seluruh item unik dari 60 transaksi (12 item), lalu transform(transactions) mengubah setiap transaksi menjadi array boolean. Nilainya True jika item dibeli pada transaksi tersebut dan False jika tidak. Array hasilnya (te_ary) dimasukkan ke dalam DataFrame dengan nama kolom dari te.columns_, sehingga setiap kolom mewakili satu item. Terakhir, df.shape dan df.iloc[:5, :6] dipakai untuk memeriksa ukuran matriks serta menampilkan lima baris dan enam kolom pertama.


Penjelasan output. Ukuran matriks adalah (60, 12), yaitu 60 baris untuk transaksi dan 12 kolom untuk item. Pada cuplikan, baris 0 dan 1 bernilai True pada kolom Gula dan Kopi, sesuai isi T1 dan T2 (Gula, Kopi, Susu). Kolom Susu tidak tampil karena cuplikan hanya memuat enam kolom pertama. Baris 2 bernilai True pada Mentega (T3), dan baris 3 bernilai True pada Mie Instan (T4, yang hanya membeli satu item). Baris 4 bernilai True pada Keju (T5). Matriks ini menjadi input langsung untuk apriori() dan fpgrowth() pada langkah berikutnya.


```python
# Menentukan minimum support
MIN_SUP = 0.15

# Menerapkan algoritma Apriori
freq_apriori = apriori(
    df,
    min_support=MIN_SUP,
    use_colnames=True
)

# Mengurutkan berdasarkan nilai support tertinggi
freq_apriori = freq_apriori.sort_values(
    'support',
    ascending=False
)

# Menampilkan hasil frequent itemsets
print('Jumlah frequent itemset:', len(freq_apriori))
print(freq_apriori.to_string(index=False))
```

Ambang batas minimum support ditetapkan MIN_SUP = 0.15, artinya sebuah itemset dianggap frequent jika muncul pada minimal 15% transaksi, atau sedikitnya 9 dari 60 transaksi. Fungsi apriori() dijalankan pada matriks one-hot df dengan parameter use_colnames=True agar itemset ditampilkan dengan nama item (misalnya Roti, Susu), bukan nomor kolom. Algoritma bekerja bertahap: pertama mencari 1-itemset yang frequent, lalu menggabungkannya menjadi kandidat 2-itemset, 3-itemset, dan seterusnya. Kandidat yang mengandung subset infrequent langsung dibuang berdasarkan prinsip downward closure. Hasilnya diurutkan dengan sort_values('support', ascending=False) supaya itemset yang paling populer berada di atas, dan to_string(index=False) dipakai agar tabel tampil rapi tanpa indeks.


Penjelasan output. Ditemukan 17 frequent itemset dengan rincian sebagai berikut.

2-itemset	6	{Roti, Susu} (0,217), {Mentega, Roti} (0,217), {Mentega, Susu} (0,183), {Mie Instan, Telur} (0,167), {Kecap, Mie Instan} (0,150), {Sabun, Shampo} (0,150)

3-itemset	1	{Mentega, Roti, Susu} (0,183)


Item Keju (0,100) dan Teh (0,067) tidak muncul karena support-nya di bawah 0,15. Tiga pola terlihat dari itemset multi-item ini. Pertama, kelompok sarapan, yaitu Mentega, Roti, dan Susu, yang muncul sebagai pasangan sekaligus satu itemset tiga item. Kedua, kelompok mie instan, yaitu Mie Instan dengan Telur dan Kecap. Ketiga, kelompok mandi, yaitu Sabun dan Shampo. Pola-pola ini diperjelas pada tahap pembentukan aturan asosiasi berikutnya.


```python
# Menentukan minimum confidence
MIN_CONF = 0.5

# Membentuk aturan asosiasi
rules = association_rules(
    freq_apriori,
    metric='confidence',
    min_threshold=MIN_CONF
)

# Mengurutkan aturan berdasarkan nilai lift tertinggi
rules = rules.sort_values(
    'lift',
    ascending=False
).reset_index(drop=True)

# Menampilkan jumlah aturan asosiasi
print('Jumlah aturan asosiasi:', len(rules))

# Memilih kolom yang akan ditampilkan
cols = [
    'antecedents',
    'consequents',
    'support',
    'confidence',
    'lift'
]

# Menampilkan hasil aturan asosiasi
print(rules[cols].round(3).to_string(index=False))
```

Setelah frequent itemset ditemukan, tahap berikutnya adalah membentuk aturan asosiasi berbentuk X → Y. Ambang batas minimum confidence ditetapkan MIN_CONF = 0.5, artinya hanya aturan yang minimal 50% transaksi pemuat X juga memuat Y yang dipertahankan. Fungsi association_rules() menerima tabel freq_apriori dengan parameter metric='confidence' dan min_threshold=MIN_CONF. Fungsi ini memecah setiap itemset yang berukuran dua item atau lebih menjadi pasangan antecedent (sisi X) dan consequent (sisi Y), lalu menghitung support, confidence, dan lift tiap aturan. Itemset tunggal tidak menghasilkan aturan karena tidak bisa dipecah. Hasilnya diurutkan berdasarkan Lift tertinggi dengan sort_values('lift', ascending=False), dan reset_index(drop=True) menyusun ulang nomor indeks. Variabel cols dipakai untuk menampilkan lima kolom utama saja, dibulatkan tiga desimal.

```python
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, association_rules

# Daftar nilai minimum support yang akan diuji
sweep_sup = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40]

# Menyimpan jumlah frequent itemsets dan aturan asosiasi
jumlah_itemset = []
jumlah_aturan = []

# Menguji setiap nilai minimum support
for sup in sweep_sup:
    fi = apriori(
        df,
        min_support=sup,
        use_colnames=True
    )

    # Membentuk aturan asosiasi
    if len(fi) > 0:
        r = association_rules(
            fi,
            metric='confidence',
            min_threshold=0.5
        )
    else:
        r = pd.DataFrame()

    # Menyimpan jumlah hasil
    jumlah_itemset.append(len(fi))
    jumlah_aturan.append(len(r))

    print(
        f'min_support={sup:.2f}: '
        f'{len(fi):2d} itemset, '
        f'{len(r):2d} aturan'
    )

# Membuat grafik pengaruh minimum support
plt.figure(figsize=(9, 5))

plt.plot(
    sweep_sup,
    jumlah_itemset,
    marker='o',
    color='#202B68',
    label='Jumlah Frequent Itemset'
)

plt.plot(
    sweep_sup,
    jumlah_aturan,
    marker='s',
    color='#C94C4C',
    label='Jumlah Aturan Asosiasi'
)

# Menandai minimum support terpilih
plt.axvline(
    x=0.15,
    color='gray',
    linestyle=':',
    linewidth=1
)

plt.annotate(
    'min_support\nterpilih = 0.15',
    xy=(0.15, 17),
    xytext=(0.17, 11),
    fontsize=8,
    color='#202B68',
    arrowprops=dict(arrowstyle='->', color='gray')
)

# Pengaturan grafik
plt.title('Pengaruh min_support terhadap Jumlah Itemset & Aturan')
plt.xlabel('min_support')
plt.ylabel('Jumlah')
plt.xticks(sweep_sup)
plt.ylim(-1, 22)
plt.grid(alpha=0.2)
plt.legend()
plt.tight_layout()
plt.show()
```

Eksperimen ini bertujuan melihat bagaimana nilai min_support memengaruhi jumlah itemset dan aturan yang ditemukan. Daftar sweep_sup berisi tujuh nilai yang diuji, yaitu 0,10 sampai 0,40 dengan selisih 0,05. Untuk setiap nilai, apriori() dijalankan pada matriks one-hot df. Jika ada itemset yang ditemukan (len(fi) > 0), association_rules() membentuk aturan dengan min_confidence = 0.5 yang tetap. Jika tidak ada itemset, r diisi DataFrame kosong supaya tidak terjadi galat, karena association_rules() tidak bisa dijalankan pada tabel kosong. Jumlah itemset dan aturan setiap iterasi disimpan ke list jumlah_itemset dan jumlah_aturan, lalu dicetak dengan format rapi.



Semakin tinggi min_support, semakin sedikit itemset dan aturan yang lolos. Pada 0,10 terdapat 19 itemset (dua lebih banyak daripada pada 0,15, yaitu Keju dan Teh? Bukan: yang bertambah hanya Keju sebagai item tunggal dan satu itemset tambahan, lihat catatan di bawah), dan jumlah aturannya naik menjadi 20. Kalau mau lanjut, no. 2 (leverage dan conviction) sudah ada outputnya di notebook kamu: Telur → Mie Instan leverage 0.0917 dan conviction 2.100, sedangkan Roti → Susu leverage 0.1217 dan conviction 2.460.

2. Untuk kedua aturan tersebut, hitung leverage dan conviction (lihat kembali rumus pada materi
perkuliahan Bagian 6), lalu bandingkan dengan Lift-nya.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# 1. Muat data (lewati kalau df sudah ada)
df_raw = pd.read_csv('transactions_arm_praktikum.csv')
transactions = df_raw['Items'].apply(lambda x: [i.strip() for i in x.split(',')]).tolist()
te = TransactionEncoder()
df = pd.DataFrame(te.fit(transactions).transform(transactions), columns=te.columns_)

# 2. Bagian 4.3-4.4: Apriori dan aturan
freq_apriori = apriori(df, min_support=0.15, use_colnames=True)
rules = association_rules(freq_apriori, metric='confidence', min_threshold=0.5)

# 3. Tugas 2: leverage dan conviction
def fs(x): return ', '.join(sorted(x))
pilih = [('Telur', 'Mie Instan'), ('Roti', 'Susu')]

for a, b in pilih:
    r = rules[(rules['antecedents'].apply(fs) == a) & (rules['consequents'].apply(fs) == b)].iloc[0]
    lev  = r['support'] - r['antecedent support'] * r['consequent support']
    conv = (1 - r['consequent support']) / (1 - r['confidence'])
    print(f'{a} -> {b}')
    print(f"  Lift       = {r['lift']:.3f}")
    print(f"  Leverage   = {r['leverage']:.4f}  (manual: {lev:.4f})")
    print(f"  Conviction = {r['conviction']:.4f}  (manual: {conv:.4f})")
```

Rumus yang digunakan:

Leverage(X→Y) = support(X→Y) − support(X) × support(Y)
Conviction(X→Y) = (1 − support(Y)) / (1 − confidence(X→Y))

Aturan 1: Telur → Mie Instan

Support = 0.167, Confidence = 0.667
Support(Telur) × Support(Mie Instan) = 0.250 × 0.300 = 0.075
Leverage = 0.167 − 0.075 = 0.0917
Conviction = (1 − 0.300) / (1 − 0.667) = 2.100
Lift = 2.222

Aturan 2: Roti → Susu

Support = 0.217, Confidence = 0.722
Support(Roti) × Support(Susu) = 0.300 × 0.317 = 0.095
Leverage = 0.217 − 0.095 = 0.1217
Conviction = (1 − 0.317) / (1 − 0.722) = 2.460
Lift = 2.281

Nilai leverage dan conviction hasil perhitungan manual sama dengan hasil mlxtend pada output kode, sehingga perhitungannya terverifikasi.

Perbandingan dengan Lift

Lift kedua aturan hampir sama (2.222 dan 2.281), sehingga jika hanya melihat Lift, keduanya tampak setara. Padahal Lift hanya membandingkan rasio dan tidak menunjukkan seberapa besar dampak pola tersebut pada seluruh transaksi. Leverage dan conviction memperlihatkan perbedaan yang tidak terlihat dari Lift.

Leverage Roti → Susu (0.1217) lebih besar dibanding Telur → Mie Instan (0.0917). Dari 60 transaksi, Roti dan Susu muncul bersama sekitar 7 transaksi lebih banyak dari yang diharapkan jika keduanya independen, sedangkan Telur dan Mie Instan sekitar 5,5 transaksi. Conviction Roti → Susu (2.460) juga lebih tinggi dibanding Telur → Mie Instan (2.100). Nilai conviction di atas 1 berarti aturan lebih jarang keliru dibanding jika kedua item independen, dan hal ini sejalan dengan confidence Roti → Susu yang lebih tinggi (0.722 dibanding 0.667).

Ketiga ukuran sama-sama menunjukkan hubungan positif (Lift > 1, leverage > 0, conviction > 1) dan sama-sama menempatkan Roti → Susu sedikit lebih kuat. **Rekomendasi Bisnis dari Aturan dengan Lift Tertinggi**

Rekomendasi berikut disusun dari aturan dengan Lift tertinggi pada dataset kedai kopi (min_support = 0.08, min_confidence = 0.5).

**1. Paket "Kopi dan Dessert": Cappuccino + Cheesecake**

Aturan Cheesecake → Cappuccino dan Cappuccino → Cheesecake memiliki Lift tertinggi (3,667), dengan confidence 0,733 dan 0,688. Pasangan ini muncul pada 11 dari 80 transaksi (support 0,138). Artinya, sekitar 73% pembeli Cheesecake juga membeli Cappuccino, dan sekitar 69% pembeli Cappuccino juga membeli Cheesecake. Kedai dapat menawarkan paket bundling Cappuccino + Cheesecake dengan harga lebih hemat, menempatkan etalase Cheesecake dekat mesin kopi atau kasir, dan meminta kasir menawarkan Cheesecake saat pelanggan memesan Cappuccino (dan sebaliknya).

**2. Paket Croissant + Latte, dengan tawaran Latte pada pembeli Americano dan Croissant**

Aturan Americano, Croissant → Latte memiliki Lift 3,386 dengan confidence 0,889, yang berarti hampir 89% pelanggan yang memesan Americano dan Croissant sekaligus juga memesan Latte. Pasangan Croissant dan Latte juga merupakan pasangan yang paling sering muncul (support 0,212, sekitar 17 dari 80 transaksi) dengan Lift 2,698. Kedai dapat membuat paket "Croissant + Latte" sebagai menu sarapan atau camilan, serta menyiapkan skrip kasir untuk menawarkan Latte kepada pelanggan yang memesan Croissant. Karena support pasangan ini tinggi, rekomendasi ini berdampak pada jumlah transaksi yang paling banyak dibanding rekomendasi lain.

**3. Paket Roti Bakar + Teh Tarik, dan promosi Roti Bakar kepada pembeli Teh Tarik**

Aturan Roti Bakar → Teh Tarik memiliki confidence 1,000 dan Lift 3,200, sehingga seluruh 12 transaksi yang memuat Roti Bakar juga memuat Teh Tarik. Kedai dapat menjadikan keduanya satu paket tetap (misalnya paket sore atau paket tradisional). Arah sebaliknya jauh lebih lemah: Teh Tarik dibeli pada 25 transaksi, tetapi hanya 12 di antaranya yang disertai Roti Bakar (confidence sekitar 0,48). Artinya masih ada sekitar 13 transaksi pembeli Teh Tarik yang belum membeli Roti Bakar. Kasir dapat menawarkan Roti Bakar kepada pembeli Teh Tarik sebagai peluang penambahan penjualan.

**4. Uji coba kecil: Air Mineral + Brownies**

Aturan Air Mineral → Brownies memiliki Lift 2,333 dan confidence 0,583, tetapi hanya didukung sekitar 7 transaksi (support 0,088). Karena dukungannya paling tipis, aturan ini sebaiknya diuji dalam skala kecil terlebih dahulu, misalnya penawaran tambahan Brownies saat pelanggan memesan Air Mineral, sebelum dijadikan paket permanen.

**Catatan**

Americano dan Teh Tarik adalah menu terpopuler (masing-masing 25 transaksi), tetapi tidak muncul pada aturan Lift tertinggi (aturan Brownies → Americano hanya ber-Lift 1,600). Kedua menu ini sudah laku dengan sendirinya sehingga lebih cocok berperan sebagai menu penarik pelanggan, sedangkan upaya cross-selling lebih efektif difokuskan pada pasangan menu di atas. Seluruh rekomendasi didasarkan pada korelasi, bukan hubungan sebab-akibat, dan dataset hanya terdiri dari 80 transaksi, sehingga dampaknya perlu divalidasi lebih lanjut, misalnya melalui A/B testing atau uji coba promo pada periode tertentu.
